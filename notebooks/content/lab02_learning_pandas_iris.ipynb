## Part 2: Exploring & cleaning data (≈10 min)

Real data has gaps. Two questions come first: **where** are the missing values, and **how** do I repair them?

- `df.isnull()` returns a True/False mask; chaining `.sum()` collapses it to a per-column **count** of gaps.
- For numeric columns, filling with the **column mean** (`df[col].fillna(df[col].mean())`) keeps the dtype numeric and barely moves the statistics. `mean()` itself skips `NaN` automatically.
- Work on a **copy** (`df.copy()`) so the original survives to compare strategies.
- Wrap cleaning in a **function** so you can reapply it to fresh data — and so Problem 4.2 can reuse it.

````python
# --- Worked example: find and repair one missing value ---
df_broken = df.copy()
df_broken.loc[0, "sepal_length"] = float("nan")   # inject one gap on purpose

missing = df_broken.isnull().sum()                 # True counts as 1 -> per-column totals
print("Missing per column:\n", missing)
print("Total missing:", int(missing.sum()))

mean_value = df_broken["sepal_length"].mean()      # mean() skips NaN automatically
df_fixed = df_broken.copy()
df_fixed["sepal_length"] = df_fixed["sepal_length"].fillna(mean_value)
print("\nAfter filling with the mean:", int(df_fixed.isnull().sum().sum()), "missing values left")
print("Original untouched:", int(df_broken.isnull().sum().sum()), "missing (we worked on copies)")
````

## Part 3: Statistics: mean, median, groupby (≈10 min)

Summary statistics turn 150 rows into a handful of insight-bearing numbers.

- `df.mean()` / `df.median()` reduce **each numeric column** to one number, returned as a Series.
- Median vs mean: when the median sits far from the mean, suspect outliers or skew.
- **Split-apply-combine:** `df.groupby("species").mean()` splits the rows by species, applies the mean within each group, and combines the results into one table — the single most useful pandas pattern for labeled data.

````python
# --- Worked example: whole-table stats vs per-species stats ---
numeric_cols = df.select_dtypes(include=[np.number]).columns
print("Numeric columns:", list(numeric_cols))

means = df[numeric_cols].mean()
medians = df[numeric_cols].median()
print("\nOverall means:\n", means.round(3))
print("\nOverall medians:\n", medians.round(3))

# Split-apply-combine: the average flower, per species
by_species = df.groupby("species")[numeric_cols].mean()
print("\nPer-species means:\n", by_species.round(3))
print("\nPetal length grows setosa < versicolor < virginica:",
      bool(by_species.loc["setosa", "petal_length"] < by_species.loc["versicolor", "petal_length"]
           < by_species.loc["virginica", "petal_length"]))
````

## Part 4: Correlation & visualization (≈10 min)

**Correlation** answers "which measurements move together?" The Pearson coefficient runs from −1 (perfect opposite) through 0 (unrelated) to +1 (perfect lockstep).

- `df[numeric].corr()` produces the full pairwise matrix — always **symmetric**, always **1s on the diagonal**.
- In iris, petal length and petal width correlate strongly (> 0.9): one measurement nearly predicts the other.
- **Histograms** reveal distribution shape: petal measurements split into visible clusters (the species!), while sepal width looks like a single bell curve.
- A heatmap (`sns.heatmap`) turns the correlation matrix into an at-a-glance picture — optional flair, same information.

````python
# --- Worked example: correlation matrix + heatmap + one histogram ---
import seaborn as sns

corr = df.select_dtypes(include=[np.number]).corr()
print("Correlation matrix:\n", corr.round(3))
print("Symmetric?", bool(np.allclose(corr, corr.T)), "| diagonal all ones?", bool(np.allclose(np.diag(corr), 1)))

plt.figure(figsize=(4.5, 3.5))
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Iris feature correlation")
plt.tight_layout()
plt.show()

df["petal_length"].plot.hist(bins=15, title="Petal length distribution")
plt.xlabel("petal length (cm)")
plt.show()
````
